```python
df_postings = spark.read.parquet("abfss://lakehouse@projectlinkedinjobs.dfs.core.windows.net/processed/postings/")
df_skills = spark.read.parquet("abfss://lakehouse@projectlinkedinjobs.dfs.core.windows.net/processed/skills/")
df_summary = spark.read.parquet("abfss://lakehouse@projectlinkedinjobs.dfs.core.windows.net/processed/summary/")

# VERIFY COUNTS
print("Postings:", df_postings.count())
print("Skills  :", df_skills.count())
print("Summary :", df_summary.count())
```

```python
silver_joined = (
    df_postings
        .join(df_skills,  "job_link", "left")
        .join(df_summary, "job_link", "left")
)
```

```python
print(f"Joined records: {silver_joined.count()}")
silver_joined.printSchema()
silver_joined.show(5, truncate=False)
```

```python
from pyspark.sql.functions import col, count, when

# Check NULL counts
null_counts = silver_joined.select([
    count(when(col(c).isNull(), c)).alias(c) 
    for c in silver_joined.columns
])
null_counts.show()
```
