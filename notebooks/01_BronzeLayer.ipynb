```python
import pandas as pd
from pathlib import Path

base = Path("/home/azureuser/kaggle_data/jobs_and_skills/files")

# Read summary CSV, skipping only truly bad rows
input_path = base / "job_summary.csv"
df = pd.read_csv(
    input_path,
    header=0,
    engine="python",
    on_bad_lines="skip"
)

# Keep only the needed columns (first 2: link + summary)
df = df.iloc[:, :2]
print("Rows:", len(df), "Columns:", df.columns.tolist())

# Write Parquet locally
parquet_path = base / "job_summary.parquet"
df.to_parquet(parquet_path, index=False)
print("Parquet saved to:", parquet_path)
```
